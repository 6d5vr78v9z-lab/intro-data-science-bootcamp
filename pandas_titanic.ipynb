### Ejercicio 8 🌶️ — Media de `age` con NumPy

Calculamos la media de la columna `age` usando NumPy (ignorando NaN con `np.nanmean`)
y la comparamos con el resultado de pandas.

```python
mean_age_numpy = np.nanmean(df['age'].values)
mean_age_pandas = df['age'].mean()

print(f'Media con np.nanmean: {mean_age_numpy}')
print(f'Media con pandas:     {mean_age_pandas}')
print(f'¿Coinciden los valores? {mean_age_numpy == mean_age_pandas}')
```

### Ejercicio 9 — Edad máxima y mínima

### Ejercicio 10 🌶️ — Nombre de la persona con la edad mínima

Usamos `idxmin()` para obtener el índice del valor mínimo y luego `.loc[]`
para localizar la fila completa.

```python
min_age_index = df['age'].idxmin()
youngest_person_name = df.loc[min_age_index, 'name']

print(f'La persona con la edad mínima es {youngest_person_name} con {df.loc[min_age_index, "age"]:.0f} años.')
```

## BONUS — Exploración del dataset del Titanic

Cargamos el dataset del Titanic directamente desde una URL pública para no
subir ningún CSV al repositorio.

```python
TITANIC_CSV_URL = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic_df = pd.read_csv(TITANIC_CSV_URL)
titanic_df.head()
```

### B1 — ¿Qué porcentaje de pasajeros sobrevivió?

```python
survival_rate_pct = titanic_df['Survived'].mean() * 100
print(f'El porcentaje de pasajeros que sobrevivió fue del {survival_rate_pct:.2f}%.')
```

### B2 — ¿Cuál es la edad media de los pasajeros?

```python
mean_passenger_age = titanic_df['Age'].mean()
print(f'La edad media de los pasajeros era de {mean_passenger_age:.2f} años.')
```

### B3 — ¿Cuántos pasajeros había en cada clase?

```python
passengers_per_class = titanic_df['Pclass'].value_counts().sort_index()
print(passengers_per_class)
print()
for pclass, count in passengers_per_class.items():
    print(f'En {pclass}ª clase viajaban {count} pasajeros.')
```

### B4 — Edad media de los pasajeros que sobrevivieron y de los que no

```python
mean_age_by_survival = titanic_df.groupby('Survived')['Age'].mean()
print(mean_age_by_survival)
print()
print(f'La edad media de los supervivientes fue de {mean_age_by_survival[1]:.2f} años.')
print(f'La edad media de los que no sobrevivieron fue de {mean_age_by_survival[0]:.2f} años.')
```

### B5 — Edad máxima de los pasajeros que sobrevivieron y de los que no

```python
max_age_by_survival = titanic_df.groupby('Survived')['Age'].max()
print(max_age_by_survival)
print()
print(f'La edad máxima entre los supervivientes fue de {max_age_by_survival[1]:.0f} años.')
print(f'La edad máxima entre los que no sobrevivieron fue de {max_age_by_survival[0]:.0f} años.')
```

### Extras — Otras exploraciones

#### Tasa de supervivencia por sexo

```python
survival_rate_by_sex = titanic_df.groupby('Sex')['Survived'].mean() * 100
print(survival_rate_by_sex)
print()
for sex, rate in survival_rate_by_sex.items():
    print(f'La tasa de supervivencia de las pasajeras {sex} fue del {rate:.2f}%.')
```

#### Tasa de supervivencia por clase

```python
survival_rate_by_class = titanic_df.groupby('Pclass')['Survived'].mean() * 100
print(survival_rate_by_class)
print()
for pclass, rate in survival_rate_by_class.items():
    print(f'La tasa de supervivencia en {pclass}ª clase fue del {rate:.2f}%.')
```

#### Tarifa media por clase

```python
mean_fare_by_class = titanic_df.groupby('Pclass')['Fare'].mean()
print(mean_fare_by_class)
print()
for pclass, avg_fare in mean_fare_by_class.items():
    print(f'La tarifa media en {pclass}ª clase fue de {avg_fare:.2f} libras.')
```
